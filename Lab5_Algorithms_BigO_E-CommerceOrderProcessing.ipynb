```python
import sys
import platform
import time
import random
 
print("Python Version:", sys.version.split()[0])
print("Operating System:", platform.system())
print("Python Executable:", sys.executable)
 
LAST_NAME = "Donor"
STUDENT_ID = "TUPM-26-7923"
FAVORITE_ARTIST = "Taylor Swift"   # change this to your own favorite artist
 
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST.upper() if c.isalpha())
 
print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps
 
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts
 
def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials
```

```python
data_size = 100 + ID_SUM * 10
 
orders = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]
 
# Three input conditions made from the SAME orders
sorted_input = sorted(orders)
reverse_input = sorted(orders, reverse=True)
random_input = orders.copy()
random.seed(ID_SUM + SEED_DIGIT)
random.shuffle(random_input)
 
conditions = [
    ("Sorted", sorted_input),
    ("Reverse", reverse_input),
    ("Random", random_input)
]
 
algorithms = [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort)
]
 
print("\nDataset Size:", data_size)
print("First 10 generated orders:", orders[:10])
print("First 10 sorted input:    ", sorted_input[:10])
print("First 10 reverse input:   ", reverse_input[:10])
print("First 10 random input:    ", random_input[:10])
```

```python
all_results = {}
expected = sorted(orders)
 
for cond_name, cond_data in conditions:
    print(f"\n===== {cond_name} Input =====")
    print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>14}{'Correct':>10}")
 
    for name, function in algorithms:
        result, elapsed = measure(function, cond_data)
        sorted_list, comparisons, movements = result
        correct = sorted_list == expected
        all_results[(cond_name, name)] = (comparisons, movements, elapsed)
 
        print(
            f"{name:<18}{comparisons:>12}{movements:>12}"
            f"{elapsed:>14.8f}{str(correct):>10}"
        )
```

```python
print("\n--- Sorting Verification ---")
all_ok = True
for cond_name, cond_data in conditions:
    for name, function in algorithms:
        sorted_list, _, _ = function(cond_data)
        ok = True
        for k in range(len(sorted_list) - 1):
            if sorted_list[k] > sorted_list[k + 1]:
                ok = False
        if not ok or len(sorted_list) != data_size:
            all_ok = False
            print("FAILED:", name, "on", cond_name)
print("All 9 results correctly sorted:", all_ok)
```

```python
print("\n--- Performance Comparison (fastest per condition) ---")
for cond_name, _ in conditions:
    fastest = None
    fastest_time = None
    for name, _f in algorithms:
        t = all_results[(cond_name, name)][2]
        if fastest_time is None or t < fastest_time:
            fastest, fastest_time = name, t
    print(f"{cond_name:<8}: {fastest} ({fastest_time:.8f} s)")
```

```python
print("\n--- Big O Analysis ---")
print(f"{'Algorithm':<18}{'Best':>8}{'Average':>10}{'Worst':>8}{'Space':>8}")
print(f"{'Bubble Sort':<18}{'O(n)':>8}{'O(n²)':>10}{'O(n²)':>8}{'O(1)':>8}")
print(f"{'Selection Sort':<18}{'O(n²)':>8}{'O(n²)':>10}{'O(n²)':>8}{'O(1)':>8}")
print(f"{'Insertion Sort':<18}{'O(n)':>8}{'O(n²)':>10}{'O(n²)':>8}{'O(1)':>8}")
 
print("\n--- Best, Average, and Worst Case Observations ---")
print("Sorted input = best case")
print("  Bubble comparisons:   ", all_results[("Sorted", "Bubble Sort")][0], "(about n, stops early)")
print("  Insertion comparisons:", all_results[("Sorted", "Insertion Sort")][0], "(about n)")
print("  Selection comparisons:", all_results[("Sorted", "Selection Sort")][0], "(still about n²/2)")
 
print("Random input = average case")
print("  Bubble comparisons:   ", all_results[("Random", "Bubble Sort")][0])
print("  Selection comparisons:", all_results[("Random", "Selection Sort")][0])
print("  Insertion comparisons:", all_results[("Random", "Insertion Sort")][0])
 
print("Reverse input = worst case")
print("  Bubble swaps:   ", all_results[("Reverse", "Bubble Sort")][1], "(about n²/2)")
print("  Insertion shifts:", all_results[("Reverse", "Insertion Sort")][1], "(about n²/2)")
print("  Selection swaps: ", all_results[("Reverse", "Selection Sort")][1], "(only about n/2)")
```
