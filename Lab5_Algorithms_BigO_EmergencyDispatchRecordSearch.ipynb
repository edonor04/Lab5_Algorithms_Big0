```python
import sys
import platform
import time
import random
 
print("Python Version:", sys.version.split()[0])
print("Operating System:", platform.system())
print("Python Executable:", sys.executable)
```

```python
LAST_NAME = "Donor"
STUDENT_ID = "TUPM-26-7923"
FAVORITE_ARTIST = "Paramore"
 
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST.upper() if c.isalpha())
 
print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
original = [1000 + i * (NAME_LENGTH + SEED_DIGIT) for i in range(20)]
 
rotation = SEED_DIGIT
rotated = original[rotation:] + original[:rotation]
 
target = original[(ID_SUM + ARTIST_SUM) % len(original)]
 
print("\nOriginal Dataset:", original)
print("Rotation Point:", rotation)
print("Rotated Dataset:", rotated)
print("Target Incident:", target)
```

```python
def rotated_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
 
        comparisons += 1
        if data[low] <= data[mid]:
            # left half is sorted
            comparisons += 1
            if data[low] <= target and target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            # right half is sorted
            comparisons += 1
            if data[mid] < target and target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons
 
 
def linear_search(data, target):
    comparisons = 0
    for i, value in enumerate(data):
        comparisons += 1
        if value == target:
            return i, comparisons
    return -1, comparisons
```

```python
index, comparisons = rotated_search(rotated, target)
li, lc = linear_search(rotated, target)
 
print("\n--- Rotated Binary Search ---")
if index != -1:
    print("Incident found at index:", index)
    print("Position (1-based):", index + 1)
else:
    print("Incident not found")
print("Comparisons:", comparisons)
 
print("\n--- Linear Search ---")
if li != -1:
    print("Incident found at index:", li)
else:
    print("Incident not found")
print("Comparisons:", lc)
 
print("\nReduction:", lc - comparisons)
```

```python
missing = 1 + ID_SUM
mi, mc = rotated_search(rotated, missing)
mli, mlc = linear_search(rotated, missing)
 
print("\n--- Not-Found Test ---")
print("Target:", missing)
print("Rotated Binary -> index =", mi, "| comparisons =", mc)
print("Linear         -> index =", mli, "| comparisons =", mlc)
if mi == -1:
    print("Incident not found")
```

```python
print("\n--- Time and Space Complexity ---")
print("Rotated Binary Search: Time O(log n), Space O(1)")
print("Linear Search:         Time O(n),     Space O(1)")
```
